## 1. ¿Qué es Python?
**Python** es un lenguaje de programación de **alto nivel**: sus instrucciones resultan relativamente fáciles de leer y no requieren manejar directamente detalles internos del hardware.

Frecuentemente se le llama **interpretado**. En su implementación habitual, **CPython**, el código fuente se compila primero a *bytecode*, que después ejecuta una máquina virtual. **Alto nivel** e **interpretado** describen características diferentes.

En bioinformática podemos utilizarlo para organizar secuencias de ADN, automatizar tareas, explorar tablas experimentales y visualizar información. Hoy comenzaremos con datos pequeños y simulados.

### La lógica detrás de un programa

**Entrada** (datos) → **Proceso** (instrucciones) → **Salida** (resultado)

Ejemplo: tres lecturas experimentales simuladas `12`, `14` y `10` → calcular su promedio → mostrar `12.0`.

````python
# ENTRADA: una lista contiene nuestras tres lecturas.
lecturas_iniciales = [12, 14, 10]

# PROCESO: sum() suma y len() cuenta cuántos datos hay.
promedio_inicial = sum(lecturas_iniciales) / len(lecturas_iniciales)

# SALIDA: print() muestra el resultado.
print("Promedio:", promedio_inicial)  # Esperamos: Promedio: 12.0
````

## 3. Nuestra primera función: `print()`

### ¿Qué es?
`print()` es una **función integrada de Python** que muestra información en la salida del notebook. Es útil para comprobar que el programa funciona, observar resultados e inspeccionar variables.

### ¿Cómo se escribe?
`print(algo_que_quiero_mostrar)`

- `print` es el nombre de la función.
- Los **paréntesis `()`** contienen la información que recibe la función.
- Las **comillas** indican que escribimos un texto literal (*string*).
- En este manual, `#` marca explicaciones que no se ejecutan.

**No es lo mismo** `print("gen")` (muestra las letras `gen`) que `print(gen)` (busca el valor de la variable llamada `gen`).

````python
# print() muestra un texto entre comillas tal como lo escribimos.
print("¡Hola, bioinformática!")

# Podemos mostrar números sin comillas.
print(42)

# Podemos mostrar varios elementos separados por comas.
print("Número de muestras:", 3)

# El texto dentro de comillas NO se interpreta como nombre de variable.
print("gen")  # Muestra literalmente: gen
````

## 4. Variables: cajas con nombre

Una **variable** es un nombre asociado a un valor. Imagínala como una caja con etiqueta: la etiqueta nos ayuda a encontrar lo que hemos guardado.

### ¿Cómo se crea?
`nombre_variable = valor`

El signo `=` **asigna** un valor: no es la comparación matemática “igual que” (`==`).

**Consejos:** usa nombres descriptivos (`numero_muestras`), evita espacios, no comiences con números y respeta mayúsculas/minúsculas (`gen` no es `Gen`).

````python
# Creamos una variable llamada gen que guarda texto.
gen = "BRCA1"

# Creamos otra que guarda un número decimal simulado.
expresion = 12.5

# Si escribimos el nombre SIN comillas, Python busca su contenido.
print("Gen:", gen)  # Gen: BRCA1
print("Expresión simulada:", expresion)  # Expresión simulada: 12.5

# Reasignamos: la variable mantiene su nombre, pero ahora guarda otro valor.
expresion = 13.2
print("Nuevo valor:", expresion)  # Nuevo valor: 13.2
````

## 5. Tipos de datos: ¿qué podemos guardar?

El **tipo de dato** nos dice qué clase de información contiene un valor.

| Tipo | Nombre | Ejemplo | Uso posible |
|---|---|---|---|
| `int` | Entero | `4` | Número de muestras |
| `float` | Decimal | `12.5` | Una lectura numérica |
| `str` | Texto | `"BRCA1"` | Nombre de un gen |
| `bool` | Booleano | `True` o `False` | Una condición sí/no |

`type(valor)` muestra el tipo de un dato. Las comillas cambian el significado: `"4"` es texto, no el número entero `4`.

````python
numero_muestras = 4        # int: número entero.
lectura = 12.5             # float: número decimal.
nombre_gen = "TP53"        # str: cadena de texto.
tiene_datos = True         # bool: verdadero o falso (sin comillas).

print(type(numero_muestras))
print(type(lectura))
print(type(nombre_gen))
print(type(tiene_datos))
````

## 6. Operadores: transformar información

Python permite hacer cálculos con `+` (sumar), `-` (restar), `*` (multiplicar), `/` (dividir), `**` (potencia).

Para comparar valores utilizamos `>` (mayor que), `<` (menor que), `>=`, `<=`, `==` (igualdad) y `!=` (diferencia). Las comparaciones devuelven `True` o `False`.

**Ejemplo de bioinformática:** calcular el promedio de varias lecturas **simuladas**, sin interpretarlas todavía como un resultado biológico.

````python
a = 12.5
b = 14.2

# Operaciones con números:
print("Suma:", a + b)
print("Promedio de dos lecturas:", (a + b) / 2)

# Comparaciones: el resultado es True o False.
print("¿a es mayor que b?", a > b)
print("¿a es igual que b?", a == b)
````

## 7. Listas: muchos datos, una sola variable

Una **lista** permite guardar varios elementos **ordenados** en una misma variable. Se escribe utilizando **corchetes `[ ]`** y comas entre elementos.

Sintaxis: `nombre = [dato1, dato2, dato3]`

### Índices: ¿cómo accedemos a los elementos?
Python comienza a contar desde **cero**: primer elemento = índice `0`; segundo = `1`; tercero = `2`.

**Analogía:** una gradilla de tubos numerada 0, 1 y 2. Cada posición guarda una lectura.

- `lista[0]`: primer elemento.
- `len(lista)`: cantidad de elementos.
- `lista.append(nuevo_valor)`: agrega un elemento al final.

````python
# Una lista de lecturas de expresión ficticias.
expresiones = [12.5, 14.2, 11.8]

# La primera posición siempre es índice 0.
print("Primera:", expresiones[0])   # 12.5
print("Segunda:", expresiones[1])   # 14.2

# len() cuenta elementos; sum() suma los números.
print("Cantidad:", len(expresiones))
print("Suma:", sum(expresiones))
print("Promedio:", sum(expresiones) / len(expresiones))

# append() incorpora una nueva lectura al final de la lista.
expresiones.append(13.1)
print("Lista actualizada:", expresiones)
````

## 8. Diccionarios: claves y valores

Un **diccionario** organiza información en pares **clave: valor**, separados por comas y encerrados entre llaves `{ }`.

Sintaxis: `muestra = {"clave": valor, "otra_clave": otro_valor}`

Se parece a una **ficha de laboratorio**: en lugar de preguntarnos “¿qué hay en la posición 1?”, podemos pedir el valor correspondiente a la etiqueta `"gen"`.

Para consultar una clave: `muestra["gen"]`. Esta búsqueda es diferente de acceder a una lista por índice.

````python
# Diccionario que describe una muestra ficticia.
muestra = {
    "id": "M01",
    "gen": "BRCA1",
    "grupo": "control",
    "expresion": 12.5
}

# Accedemos mediante el nombre de la clave.
print("Identificador:", muestra["id"])
print("Gen:", muestra["gen"])
print("Grupo:", muestra["grupo"])
````

## 9. Condicionales: tomar decisiones con `if`

Un **condicional** permite ejecutar distintas instrucciones dependiendo de si una condición es verdadera o falsa.

Sintaxis general:

```python
if condicion:
    # Instrucciones si la condición es verdadera.
else:
    # Instrucciones si es falsa.
```

- `if` significa **si**.
- `elif` permite revisar otra condición, si la primera no se cumplió.
- `else` significa **en otro caso**.
- Los **dos puntos `:`** anuncian que empieza un bloque.
- La **indentación** (espacios al comienzo de la línea) indica qué instrucciones pertenecen al bloque. Habitualmente son **4 espacios**.

**Advertencia:** la condición del ejemplo es solo una regla didáctica, no un umbral validado de expresión génica.

````python
# Definimos un valor ficticio y un umbral inventado para practicar.
expresion_ejemplo = 12.5
umbral_didactico = 10

# Si se cumple la comparación, se ejecuta el bloque indentado.
if expresion_ejemplo > umbral_didactico:
    print("Supera el umbral del ejercicio")
else:
    print("No supera el umbral del ejercicio")
````

## 11. Ciclos o bucles `for`: repetir instrucciones

Un **bucle** permite recorrer varios elementos sin escribir manualmente la misma instrucción una y otra vez.

Sintaxis:
```python
for elemento in coleccion:
    # Instrucciones que se repiten para cada elemento.
```

La variable `elemento` toma sucesivamente los valores de la colección. **Analogía científica:** revisar uno por uno los tubos de una gradilla.

````python
# Tenemos una lista de nombres de genes como ejemplo didáctico.
genes = ["BRCA1", "TP53", "GAPDH"]

# En cada vuelta, gen_actual toma el siguiente valor de la lista.
for gen_actual in genes:
    print("Estamos revisando:", gen_actual)

# Se ejecuta una vez por cada elemento: aquí, tres veces.
````

### ¿Y `range()`?
`range(n)` genera una secuencia de enteros que empieza en `0` y termina **antes** de `n`. Es útil cuando queremos repetir una tarea una cantidad concreta de veces.

````python
# range(3) produce 0, 1 y 2.
for numero in range(3):
    print("Iteración:", numero)
````

## 12. Funciones: recetas reutilizables

Una **función** es un bloque de instrucciones que podemos ejecutar cuando lo necesitemos. Nos ayuda a no repetir código, mantener el trabajo organizado y reutilizar procedimientos.

- `def` inicia la definición de una función.
- Entre paréntesis ponemos los **parámetros** (datos que espera recibir).
- `return` devuelve un resultado para que podamos utilizarlo después.

**Analogía:** una receta que recibe ingredientes (entrada), aplica pasos (proceso) y entrega un resultado (salida).

`print()` **muestra** algo, mientras que `return` **devuelve** un valor. Son cosas distintas.

````python
# Creamos una función que recibe una lista de números.
def calcular_promedio(valores):
    # La variable resultado existe dentro de la función.
    resultado = sum(valores) / len(valores)
    # return devuelve el número calculado.
    return resultado

# Llamamos la función con datos ficticios.
resultado_funcion = calcular_promedio([12.0, 14.0, 10.0])
print("Promedio calculado:", resultado_funcion)
````

## 14. ¿Qué es un DataFrame?

Un **DataFrame** es una estructura de datos de Pandas organizada en **filas y columnas**, parecida a una hoja de cálculo.

- **Fila:** representa un registro o una observación (por ejemplo, una muestra).
- **Columna:** representa una característica (por ejemplo, grupo, gen o expresión).
- **Valor:** el dato que se encuentra en la intersección entre fila y columna.

En bioinformática, esta organización facilita explorar tablas de resultados antes de hacer cualquier análisis más avanzado.

````python
# Primero creamos un diccionario: cada clave será una columna.
datos_pequenos = {
    "muestra": ["M01", "M02", "M03"],
    "gen": ["BRCA1", "TP53", "GAPDH"],
    "expresion": [12.5, 9.8, 20.1]
}

# pd.DataFrame() transforma el diccionario en una tabla.
tabla_pequena = pd.DataFrame(datos_pequenos)

# display() presenta el DataFrame como tabla en Colab.
display(tabla_pequena)
````

## 15. ¿Qué es un archivo CSV?

**CSV** significa *Comma-Separated Values* (valores separados por comas). Es un archivo de texto que guarda información en filas y columnas; por eso es muy común para intercambiar datos.

Ejemplo de archivo CSV:
```text
muestra,gen,expresion
M01,BRCA1,12.5
M02,TP53,9.8
```

Para leerlo usamos `pd.read_csv("nombre_archivo.csv")`.

## 16. Explorar antes de modificar: `head()`, `info()` y `describe()`

| Instrucción | ¿Qué hace? | Pregunta que responde |

## 18. ¿Qué son los valores faltantes (`NaN`)?

En tablas reales puede haber celdas sin información. Pandas suele representarlas como **`NaN`** (*Not a Number*) en columnas numéricas.

**Una celda faltante no significa que el valor sea cero.** No debemos inventar un dato para llenar espacios. La decisión de eliminar, reemplazar o investigar faltantes corresponde a una fase posterior de análisis y depende de la pregunta científica.

### ¿Cómo identificarlos?
- `df.isna()` devuelve una tabla de `True`/`False`: ¿falta información en esta celda?
- `df.isna().sum()` cuenta los valores faltantes **en cada columna**.

````python
# isna() identifica qué celdas son faltantes.
print("¿Hay valores faltantes en cada celda?")
display(df.isna().head())

# sum() sobre booleanos cuenta True, es decir, los datos faltantes.
print("Faltantes por columna:")
print(df.isna().sum())  # En nuestros datos, expresion tiene 2 faltantes.
````

# 🛟 Guía rápida de errores frecuentes

### `NameError: name ... is not defined`
Python no conoce ese nombre. ¿Ejecutaste antes la celda que crea la variable? ¿Escribiste el nombre igual, respetando mayúsculas?

### `SyntaxError`
La instrucción no está escrita correctamente. Revisa comillas, paréntesis, corchetes, comas y dos puntos.

### `IndentationError`
Falta o sobra sangría. Revisa los espacios de las líneas dentro de `if`, `for` y `def`.

### `TypeError`
Estás combinando tipos de datos de forma incompatible. Por ejemplo, no puedes sumar directamente `"10"` (texto) con `2` (número).

### `FileNotFoundError`
El CSV no está en la carpeta esperada. Ejecuta primero la celda que crea nuestro archivo de ejemplo o sube el archivo nuevamente.

### Los datos cambiaron o desaparecieron
Colab puede reiniciar el entorno de ejecución. Vuelve a ejecutar las celdas en orden; conserva el notebook y guarda cambios en GitHub al terminar una parte importante.

### GitHub no muestra mis cambios
Guardar una copia en GitHub **no es un guardado continuo**. Repite el proceso, verifica el repositorio y busca el commit más reciente.
